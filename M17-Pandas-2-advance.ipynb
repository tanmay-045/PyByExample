### Q4. Given a Pandas DataFrame df with a column 'Text', write a Python function to create a new column 'Word_Count' that contains the number of words in each row of the 'Text' column.

````python
df= pd.DataFrame(
    {
    'Text':
        ['An Apple a day keeps the doctor away.',
         'Eating pineapple may help boost immunity.',
         'Eating watermelon promotes good health.']     
    }
)

def new_wc(df):
    lst=[]
    for i in range(len(df)):
        wc=len(df['Text'][i].split())
        lst.append(wc)
    
    df['Word_Count']=lst
    return df


new_wc(df)
````

````python
# aliter

import pandas as pd

df = pd.DataFrame({
    'Text': [
        'An Apple a day keeps the doctor away',
        'Eating pineapple may help boost immunity',
        'Eating watermelon promotes good health'
    ]
})

def count_words(text):
    # Split the text into words and count them
    return len(text.split())

# Apply the function to create the 'Word_Count' column
df['Word_Count'] = df['Text'].apply(count_words)

# Display the updated DataFrame
print(df)
````

### Q7. You have a Pandas DataFrame df that contains a column named 'Email' that contains email addresses in the format 'username@domain.com' . Write a Python function that creates a new column 'Username' in df that contains only the username part of each email address.

#### The username is the part of the email address that appears before the '@' symbol. For example, if the email address is 'john.doe@example.com', the 'Username' column should contain 'john.doe'. Your function should extract the username from each email address and store it in the new 'Username' column.

````python
df=pd.DataFrame({
        'Email' : [
                    'john.doe@example.com',
                    'margha@sample.com',
                    'janeblue@ghail.com',
                    'tanmay_sharma@newmail.com',
                    'genilia.gim@catmail.com']

})

# defining the function
def new_col_username(df):
    unames=[]
    for i in df['Email']:
        unames.append(i.split('@')[0])
    df['Username']=unames
    return df

# calling the function
new_col_username(df)
````

````python
# Aliter

df=pd.DataFrame({
        'Email' : [
                    'john.doe@example.com',
                    'margha@sample.com',
                    'janeblue@ghail.com',
                    'tanmay_sharma@newmail.com',
                    'genilia.gim@catmail.com']
})

# defining function to get username and create a new column for it
def username(df):
    # Using the .str accessor for vectorized string operations
    df['Username']=df['Email'].str.split('@').str[0]
    return df

# defining function to get domain names
def email_domain(df):
    temp=df["Email"].str.split('@').str[1]
    df['Domain_Name']=temp.str.split('.com').str[0]
    return df

print(username(df),'\n\n')
print(email_domain(df))
````

### Q10. Given a Pandas DataFrame df with a column 'Sales' and a column 'Date', write a Python function to create a new column 'MovingAverage' that contains the moving average of the sales for the past 7 days for each row in the DataFrame. The moving average should be calculated using a window of size 7 and should include the current day.


````python
# Sample DataFrame
data = {
    'Date': pd.date_range(start='2023-01-01', periods=15, freq='D'),
    'Sales': [100, 150, 200, 250, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750, 800]
}
df = pd.DataFrame(data)

# Ensure the DataFrame is sorted by date
df.sort_values(by='Date', inplace=True)

# Defining the function to calculate the 7-day moving average
def add_moving_average(df):
    df['MovingAverage'] = df['Sales'].rolling(window=7, min_periods=1).mean()
    return df

# Calling the function
df = add_moving_average(df)

# Displaying the DataFrame
print(df)
````

### Q11. You have a Pandas DataFrame df with a column 'Date'. Write a Python function that creates a new column 'Weekday' in the DataFrame. The 'Weekday' column should contain the weekday name (e.g. Monday, Tuesday) corresponding to each date in the 'Date' column.

````python
# Sample DataFrame
data = {
    'Date': pd.date_range(start='2024-07-01', periods=7, freq='D')
}
df = pd.DataFrame(data)

# Defining the function to add the 'Weekday' column
def add_weekday_column(df):
    df['Weekday'] = df['Date'].dt.day_name()
    return df

# Calling the function
df = add_weekday_column(df)

# Displaying the DataFrame
print(df)
````

### Q12. Given a Pandas DataFrame df with a column 'Date' that contains timestamps, write a Python function to select all rows where the date is between '2023-01-01' and '2023-01-31'.

**notes**
- `df.loc` is a primary attribute in pandas used for selecting rows and columns from a DataFrame by label-based indexing.
- syntax `df.loc[row_labels, column_labels]`

````python
# Sample DataFrame
data = {
    'Date': pd.date_range(start='2023-01-01', periods=50, freq='D')
}
df = pd.DataFrame(data)

# Defining the function to filter rows based on date range
def filter_by_date(df):
    start_date = '2023-01-01'
    end_date = '2023-01-31'
    mask = (df['Date'] >= start_date) & (df['Date'] <= end_date)
    filtered_df = df.loc[mask]
    return filtered_df

# Calling the function
filtered_df = filter_by_date(df)

# Displaying the filtered DataFrame
print(filtered_df)
````

## extras
The `pd.date_range()` function in pandas is used to generate a fixed-frequency DatetimeIndex. It's a versatile tool for creating date and time sequences, which is essential for working with time series data in pandas.

### Syntax:
```python
pd.date_range(start=None, end=None, periods=None, freq=None, tz=None, normalize=False, name=None, closed=None, **kwargs)
```

### Parameters:

- **`start`**: Start date for the range. If not specified, defaults to `None`.
- **`end`**: End date for the range. If not specified, defaults to `None`.
- **`periods`**: Number of periods (e.g., days) to generate. Specify either `end` or `periods`, but not both.
- **`freq`**: Frequency strings or DateOffset objects specifying the frequency of the dates to be generated. This is a required parameter if `start` and `end` are not specified.
- **`tz`**: Time zone name for the DatetimeIndex. Default is `None` (no timezone).
- **`normalize`**: If True, normalize the start and end dates to midnight before generating the date range.
- **`name`**: Name to be stored in the resulting DatetimeIndex.
- **`closed`**: Defines whether to include `start` and `end` dates. Possible values are 'right', 'left', 'both', 'neither'. Default is `None`.

### Frequencies (`freq` parameter):
The `freq` parameter allows you to specify the frequency of the dates to be generated. Some common frequency strings include:

- **`'D'`**: calendar day frequency
- **`'B'`**: business day frequency
- **`'W'`**: weekly frequency
- **`'M'`**: month end frequency
- **`'Q'`**: quarter end frequency
- **`'A'`**: year end frequency

These can be combined with numbers to specify multiples, e.g., `'3D'` for every 3 days, `'2W'` for every 2 weeks.

````python
import pandas as pd

# Generate a date range from '2023-01-01' to '2023-01-31'
date_range = pd.date_range(start='2023-01-01', end='2023-01-31')
print(date_range)
````

````python
# Generate a date range with a weekly frequency starting from '2023-01-01'
weekly_range = pd.date_range(start='2023-01-01', periods=13, freq='W')
print(weekly_range)
````
