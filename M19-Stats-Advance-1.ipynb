### Q1. What is the Probability density function?

The Probability Density Function (PDF) is a function that describes the likelihood of a random variable to take on a given value. In the context of continuous random variables, the PDF, often denoted as f(x) , provides the relative likelihood that the random variable  'X' takes on a value near x.

Here are key characteristics and properties of a PDF:

1. **Non-negativity**: The PDF is always non-negative, i.e., ' f(x) >=0 ' for all  x.

2. **Normalization**: The total area under the PDF curve is equal to 1.
   This ensures that the total probability across all possible values of the random variable is 1.

3. **Probability Calculation**: The probability that a continuous random variable X falls within an interval [a,b] is given by the area under the PDF curve between  a and b.
    
4. **Value Interpretation**: For any specific point x, f(x) itself is not a probability but rather a density. The value  f(x) can be greater than 1 for some  'x', especially when dealing with variables defined over a small range of values.

### Q2. What are the types of Probability distribution?

### 1. Discrete Probability Distributions

Discrete distributions describe variables that take on a countable number of distinct values.

- **Bernoulli Distribution**: Represents a random experiment with only two outcomes, typically "success" and "failure." The probability of success is denoted by \( p \).

- **Binomial Distribution**: Describes the number of successes in a fixed number of independent Bernoulli trials. It is defined by the number of trials \( n \) and the probability of success \( p \).

- **Geometric Distribution**: Models the number of trials until the first success in a series of Bernoulli trials. It is defined by the probability of success \( p \).

- **Poisson Distribution**: Describes the number of events occurring in a fixed interval of time or space, given the average number of occurrences \( \lambda \).

### 2. Continuous Probability Distributions

Continuous distributions describe variables that take on an infinite number of possible values within a given range.

- **Normal (Gaussian) Distribution**: Characterized by its bell-shaped curve, defined by its mean \( \mu \) and standard deviation \( \sigma \).

- **Exponential Distribution**: Models the time between events in a Poisson process. It is defined by the rate parameter \( \lambda \).

- **Uniform Distribution**: All outcomes are equally likely within a specified range \([a, b]\).

- **Gamma Distribution**: Generalizes the exponential distribution and is often used in queuing models. It is defined by shape and rate parameters.

- **Beta Distribution**: Defined on the interval \([0, 1]\) and used in Bayesian statistics. It is characterized by shape parameters \( \alpha \) and \( \beta \).


### Q3. Write a Python function to calculate the probability density function of a normal distribution with given mean and standard deviation at a given point.

![img1](z_img_pdf_formula.png)

```python
import math

def normal_pdf(x, mean, std_dev):
    """
    Calculate the probability density function of a normal distribution.
    
    Parameters:
    x (float): The point at which to evaluate the PDF.
    mean (float): The mean of the normal distribution.
    std_dev (float): The standard deviation of the normal distribution.
    
    Returns:
    float: The value of the PDF at the given point.
    """
    # Calculate the coefficient (1 / (std_dev * sqrt(2 * pi)))
    coefficient = 1 / (std_dev * math.sqrt(2 * math.pi))
    
    # Calculate the exponent ((x - mean)^2 / (2 * std_dev^2))
    exponent = -((x - mean) ** 2) / (2 * (std_dev ** 2))
    
    # Calculate the PDF value
    pdf_value = coefficient * math.exp(exponent)
    
    return pdf_value

# Example usage:
mean = 0
std_dev = 1
x = 1
pdf_value = normal_pdf(x, mean, std_dev)
print(f"The PDF value of a normal distribution at x = {x} with mean = {mean} and std_dev = {std_dev} is {pdf_value}")
```

### Q4. What are the properties of Binomial distribution? Give two examples of events where binomial distribution can be applied.


The properties of a Binomial distribution are:

1. **Two Possible Outcomes**: Each trial results in either success or failure.
2. **Fixed Number of Trials**: The number of trials \( n \) is fixed.
3. **Constant Probability**: The probability of success \( p \) is the same for each trial.
4. **Independent Trials**: Each trial is independent of the others.

### Examples of Events:

1. **Flipping a Coin**: Counting the number of heads (successes) in 10 coin flips.
2. **Quality Control**: Counting the number of defective items (failures) in a batch of 50 products.

### Q5. Generate a random sample of size 1000 from a binomial distribution with probability of success 0.4 and plot a histogram of the results using matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np

# Parameters
n = 10  # Number of trials
p = 0.4  # Probability of success
sample_size = 1000

# Generate random sample from binomial distribution
sample = np.random.binomial(n, p, sample_size)

# Plotting histogram
plt.figure(figsize=(8, 6))
plt.hist(sample, bins=np.arange(n + 2) - 0.5, alpha=0.75, edgecolor='black', linewidth=1.2)
plt.xticks(np.arange(n + 1))
plt.xlabel('Number of Successes')
plt.ylabel('Frequency')
plt.title(f'Binomial Distribution: n={n}, p={p}, Sample Size={sample_size}')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Q6. Write a Python function to calculate the cumulative distribution function of a Poisson distribution with given mean at a given point.

![img2](z_img_cdf_formula.png)

```python
import math

def poisson_cdf(k, mean):
    """
    Calculate the cumulative distribution function (CDF) of a Poisson distribution.

    Parameters:
    k (int): The point at which to evaluate the CDF.
    mean (float): The mean (lambda) of the Poisson distribution.

    Returns:
    float: The cumulative probability up to the given point.
    """
    if k < 0:
        return 0.0
    
    cumulative_prob = 0.0
    for i in range(k + 1):
        cumulative_prob += math.exp(-mean) * (mean ** i) / math.factorial(i)
    
    return cumulative_prob

# Example usage:
mean = 2.5
k = 3
cdf_value = poisson_cdf(k, mean)
print(f"The cumulative probability of a Poisson distribution with mean = {mean} at point k = {k} is {cdf_value}")
```

### Q7. How Binomial distribution different from Poisson distribution?


| Feature                      | Binomial Distribution                                 | Poisson Distribution                             |
|------------------------------|-------------------------------------------------------|--------------------------------------------------|
| **Number of Trials**         | Fixed 'n'                                             | Infinite or large 'n' not specified              |
| **Probability of Success**   | Constant  'p'                                         | Fixed rate parameter 'lambda'                    |
| **Outcome**                  | Discrete (number of successes ' k ')                 | Discrete (number of events ' k ')                |
| **Conditions**               | Requires fixed number of trials and independent       | Typically used for rare events or in the limit   |
| **Mean**                     | ' mu = n * p '                                        | mu = 'lambda'                                    |
| **Variance**                 | ' sigma^2 = n * p * (1 - p) '                         | ' sigma^2 = lambda                               |
| **Example Applications**     | Number of successes in ' n ' trials, coin flips       | Number of calls received by a call center per hour

### Q8. Generate a random sample of size 1000 from a Poisson distribution with mean 5 and calculate the sample mean and variance.

```python
import numpy as np

# Parameters
mean = 5
sample_size = 1000

# Generate random sample from Poisson distribution
sample = np.random.poisson(mean, sample_size)

# Calculate sample mean and variance
sample_mean = np.mean(sample)
sample_variance = np.var(sample)

print(f"Sample Mean: {sample_mean}")
print(f"Sample Variance: {sample_variance}")
```

### Q9. How mean and variance are related in Binomial distribution and Poisson distribution?

In both Binomial and Poisson distributions, the mean μ and variance σ^2 are closely related, but the nature of their relationship differs due to the distinct characteristics of each distribution:

### Binomial Distribution:

- **Mean μ**: For a Binomial distribution with parameters ' n ' (number of trials) and ' p ' (probability of success in each trial), the mean is given by:
   μ = n * p
  

- **Variance σ^2**: The variance of a Binomial distribution is:
  σ^2 = n * p * (1 - p)

### Poisson Distribution:

- **Mean  μ**: For a Poisson distribution with parameter λ (average rate of occurrence), the mean is:

  μ = λ
  

- **Variance '𝜎^2 '**: The variance of a Poisson distribution is also λ:
  
  𝜎^2 = λ

### Relationship:

In both distributions, the variance is directly proportional to the mean. Specifically:

- **Binomial Distribution**: The variance ' 𝜎^2 ' is proportional to both ' n ' (number of trials) and ' p ' (probability of success). As ' p ' approaches 0 or 1, the variance decreases.

- **Poisson Distribution**: The variance ' 𝜎^2 ' is equal to the mean λ. This indicates a specific relationship where the variance is equal to the mean.

### Q10. In normal distribution with respect to mean position, where does the least frequent data appear?



In a normal distribution:

- The mean (\( \mu \)) represents the center or average of the distribution.
- The least frequent data points, which are the extreme values, appear farthest from the mean.

Specifically:

- **Least Frequent Data**: These are the data points located in the tails of the normal distribution, farthest away from the mean (\( \mu \)).
- **Position**: The least frequent data points are located at the tails of the distribution, which are at the highest and lowest ends of the x-axis.

In a standard normal distribution (mean \( \mu = 0 \) and standard deviation \( \sigma = 1 \)), the least frequent data points are typically found beyond \( \pm 2 \) standard deviations from the mean.
